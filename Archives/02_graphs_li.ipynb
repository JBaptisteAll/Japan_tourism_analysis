# Slide 2
## Overall people intersted about a trip to Japan

🎯 Constater un fort intérêt ne suffit pas.

Sur ce projet, près de 65 % des répondants n’ont jamais visité le Japon…    
mais déclarent clairement vouloir y aller.

🧠 Le problème n’était donc pas la notoriété.  
L’envie était déjà là.

📊 La vraie question était ailleurs :  
pourquoi autant de désir ne se transforme pas en passage à l’action ?

👉 Avant même d’analyser plus finement,  
il a fallu reformuler le problème :  
ce n’était pas une question d’intérêt,  
mais une question de conversion.

💬 Dans vos projets data, à quel moment prenez-vous le temps de reformuler le problème avant d’ouvrir le dataset ?

```python
df_slide2 = df
```

```python
mapping_xp_japan = {
    "Yes, several times": "Déjà allés",
    "Yes, once": "Déjà allés",
    "No, and I’m not interested": "Pas intéressés (jamais allés)",
    "No, but I would like to go": "Intéressés (jamais allés)"
}

df_slide2["been_to_Japan"] = (df["been_to_Japan"].map(mapping_xp_japan))
```

```python
df_xp_japan = df_slide2["been_to_Japan"].value_counts().reset_index()
```

```python
print(df_xp_japan)
```

```python
df_xp_japan["percentage"] = (
    df_xp_japan["count"] / df_xp_japan["count"].sum() * 100
).round(2)
```

```python
print(df_xp_japan)
```

Pie chart

```python
fig = px.pie (
    df_xp_japan,
    values = "count",
    names = "been_to_Japan",
    title = "Un intérêt fort… mais une conversion faible",
    hole = 0
)

fig.update_traces (
    textinfo = "percent+label"
)

fig.update_layout (
    showlegend = True
)
```

Bar chart

```python
fig = px.bar (
    df_xp_japan,
    x = "been_to_Japan",
    y = "percentage",
    text = "percentage",
    title = "Un intérêt fort… mais une conversion faible",
    color = "count"
)

fig.update_traces (
    texttemplate = "%{text}%",
    textposition="outside"
)

fig.update_layout (
    width=800,
    height=500,
    xaxis_title = "",
    yaxis_title = "Pourcentage de répondants",
    coloraxis_showscale=False
)
```

```python
df_xp_japan["text_display"] = df_xp_japan["percentage"].astype(str) + "%"

df_xp_japan.loc[
    df_xp_japan["been_to_Japan"] == "Intéressés (jamais allés)",
    "text_display"
] = "<b>" + df_xp_japan["text_display"] + "</b>"
```

```python
fig = px.bar (
    df_xp_japan,
    x = "been_to_Japan",
    y = "percentage",
    text = "text_display",
    title = "Un intérêt fort… mais une conversion faible",
    color = "been_to_Japan",
    color_discrete_map = {
        "Déjà allés": "#0C0241",
        "Pas intéressés (jamais allés)": "#0C0241",
        "Intéressés (jamais allés)": "#E6FF01"
        
    }
)

fig.update_traces (
    textfont = dict(size=14),
    textposition = "outside"
)

fig.update_layout (
    width=800,
    height=500,
    xaxis_title = "",
    yaxis_title = "Part des répondants (%)",
    showlegend = False
)
```

# Slide 3
## SEGMENTATION

filter df to keep only those 64%

📊 Tous les voyageurs intéressés par le Japon ne veulent pas la même chose.

Après avoir constaté un paradoxe clair  
👉 un intérêt très fort, mais une conversion faible,

la question suivante était simple :  
qu’est-ce qui est réellement ESSENTIEL pour ces voyageurs ?

🧠 Pour y répondre, je n’ai pas regardé des moyennes.  
J’ai analysé la part des répondants pour qui un critère est jugé “ESSENTIEL”.

Le résultat est sans ambiguïté :  
🍜 la gastronomie arrive largement en tête  
🌿 la nature et la randonnée suivent  
🏯 la culture et le patrimoine restent centraux

À l’inverse, le shopping, les événements ou les parcs à thème  
sont perçus comme secondaires par la majorité.

👉 Même niveau d’intérêt pour le Japon.  
👉 Mais des priorités très différentes.

💬 Dans vos analyses, regardez-vous ce qui est “important”…  
ou ce qui est réellement non négociable pour vos utilisateurs ?

```python
mapping_interest = {
    "Not important at all": 0,
    "Slightly important": 1,
    "Moderately important": 2,
    "Very important": 3,
    "Essential": 4,
}
```

```python
columns = [
    'rating_interest_culture_and_history', 
    'rating_interest_food', 
    'rating_interest_nature_hiking', 
    'rating_interest_shopping_and_techno',
    'rating_interest_events_and_festivals', 
    'rating_interest_wellness',
    'rating_interest_theme_park'
]
```

```python
df_numeric = df.copy()

for col in columns:
    df_numeric[col] = df_numeric[col].map(mapping_interest)
```

```python
label_map = {
    "rating_interest_culture_and_history": "Culture & History",
    "rating_interest_food": "Food",
    "rating_interest_nature_hiking": "Nature & Hiking",
    "rating_interest_shopping_and_techno": "Shopping & Tech",
    "rating_interest_events_and_festivals": "Events & Festivals",
    "rating_interest_wellness": "Wellness",
    "rating_interest_theme_park": "Theme Parks",
}
```

```python

summary_df = (
    df_numeric[columns]
    .mean()
    .reset_index()
    .rename(columns = {
        "index": "interest",
        0: "value"
    })
)

summary_df["text_display"] = summary_df["value"].round(2)
summary_df["interest"] = summary_df["interest"].map(label_map)


fig_1 = px.bar (
    summary_df,
    x = "interest",
    y = "value",
    text = "text_display",
    title = "Un intérêt fort… mais une conversion faible",
)

fig_1.update_traces (
    textposition = "outside"
)

fig_1.update_layout (
    width=800,
    height=500,
    xaxis_title = "",
    yaxis_title = "Score moyen d’intérêt (0–4)"
)
```

```python
essential_pct = []

for col in columns:
    pct = (df_numeric[col] == 4).mean() * 100
    essential_pct.append({
        "interest": col.replace("rating_interest_", "").replace("_", " ").title(),
        "essential_pct": round(pct, 1)
    })

essential_df = pd.DataFrame(essential_pct).sort_values(
    "essential_pct", ascending=False
)

essential_df
```

```python
essential_df["text_display"] = essential_df["essential_pct"].astype(str) + "%"


essential_df = essential_df.sort_values(
    "essential_pct", ascending=False
).reset_index(drop=True)

essential_df.loc[0, "text_display"] = "<b>" + essential_df.loc[0, "text_display"] + "</b>"
```

```python
#essential_df["interest"] = essential_df["interest"].map(label_map)
mapping_interest = {
    "Food": "Gastronomie",
    "Nature Hiking": "Nature & randonnée",
    "Culture And History": "Culture & patrimoine",
    "Wellness": "Bien-être",
    "Shopping And Techno": "Shopping & technologie",
    "Events And Festivals": "Événements & festivals",
    "Theme Park": "Parcs à thème"
}

essential_df["interest"] = (essential_df["interest"].map(mapping_interest))
```

```python
fig = px.bar(
    essential_df,
    x="interest",
    y="essential_pct",
    text="text_display",
    title="Critères jugés ESSENTIELS par les voyageurs intéressés par le Japon (%)",
    color = "interest",
    color_discrete_map = {
        "Gastronomie": "#E6FF01",
        "Nature & randonnée": "#E9A107",
        "Culture & patrimoine": "#E9A107",
        "Bien-être": "#0C0241",
        "Shopping & technologie": "#0C0241",
        "Événements & festivals": "#313131",
        "Parcs à thème": "#313131"
        }
)

fig.update_traces(textfont = dict(size=15), textposition="outside")
fig.update_layout(
    width=800,
    height=500,
    xaxis_title="",
    yaxis_title="Part des répondants (%)",
    yaxis_range=[0, max(essential_df["essential_pct"]) * 1.2],
    showlegend = False
)

fig.show()
```

## FREINS STRUCTURELS & FRICTION

filter df to keep only those 64%

🚧 Si l’intérêt est fort, pourquoi la conversion reste-t-elle faible ?

Après avoir identifié ce qui motive réellement les voyageurs,  
il restait une question clé :

👉 Qu’est-ce qui freine concrètement le passage à l’action ?

🧠 L’analyse montre un résultat très clair :  
la barrière principale n’est ni le budget, ni la popularité,  
mais la LANGUE.

Et ce point est crucial :  
ce frein persiste quel que soit le niveau d’expérience.  
Même chez les personnes déjà allées au Japon.

👉 L’expérience réduit certaines frictions…  
👉 mais n’élimine pas les freins structurels.

Autrement dit :  
le problème n’est pas un manque d’envie,  
mais un manque de réassurance.

💬 Dans vos projets data, combien de “freins” identifiés  
sont en réalité structurels plutôt que temporaires ?
